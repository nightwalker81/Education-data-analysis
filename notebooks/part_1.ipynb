## Etape 2 : Réaliser le premier nettoyage

```python
data = pd.read_csv("data/EdStatsCountry.csv").dropna(how="all")
data.head()
del data["Unnamed: 31"]
```

```python
data.duplicated().sum()
```

```python
# la creation d'une liste de faux pays

fake_df = data[data["Region"].isnull()]

faux_pays = []
for val in fake_df["Short Name"]:
    faux_pays.append(val)
print(faux_pays)
```

```python
# filteration des faux pays

filtered_df = data[~data["Short Name"].isin(faux_pays)]
```

```python
# Fusionner le dataframe nettoyé avec df2 

df2_clean = df2.merge(filtered_df[["Short Name"]], on = "Short Name",how = "inner")
```

```python
# Fusionner le dataframe nettoyé avec df3

df3_clean = df3.merge(filtered_df[["Short Name"]], how = "inner", left_on = "Country Name", right_on = "Short Name")
```

```python
# Fusionner le dataframe nettoyé avec df4

df4_clean = df4.merge(filtered_df[["Country Code"]], how="inner", left_on = "CountryCode", right_on = "Country Code").drop("CountryCode", axis = "columns")
```

```python
# Fusionner le dataframe nettoyé avec df4
df1_clean = df1.merge(filtered_df[["Country Code"]], how="inner", left_on = "CountryCode", right_on = "Country Code").drop("CountryCode", axis = "columns")
```
